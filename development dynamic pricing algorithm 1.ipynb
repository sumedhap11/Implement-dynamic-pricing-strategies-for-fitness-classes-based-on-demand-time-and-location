```python
def dynamic_price(base_price, demand, popularity, time_factor, location_factor):
    # weights (can be tuned)
    alpha = 0.3   # demand weight
    beta = 0.2    # popularity weight
    gamma = 0.1   # time weight
    delta = 0.15  # location weight
    
    price = base_price * (
        1 
        + alpha * demand 
        + beta * popularity 
        + gamma * time_factor 
        + delta * location_factor
    )
    
    return round(price, 2)


# Example inputs
base_price = 500
demand = 0.8          # high demand
popularity = 0.7      # popular class
time_factor = 0.5     # peak time
location_factor = 0.6 # urban area

final_price = dynamic_price(base_price, demand, popularity, time_factor, location_factor)

print("Final Dynamic Price:", final_price)
```

```python
def pricing_rules(base_price, demand, seats_left):
    if demand > 0.8 and seats_left < 10:
        return base_price * 1.5   # surge pricing
    elif demand < 0.3:
        return base_price * 0.8   # discount
    else:
        return base_price
```

```python
import pandas as pd

# sample historical data
data = pd.DataFrame({
    'base_price': [500, 500, 500],
    'demand': [0.9, 0.4, 0.7],
    'popularity': [0.8, 0.3, 0.6],
    'time_factor': [0.6, 0.2, 0.5],
    'location_factor': [0.7, 0.4, 0.5]
})

# apply dynamic pricing
data['dynamic_price'] = data.apply(
    lambda row: dynamic_price(
        row['base_price'],
        row['demand'],
        row['popularity'],
        row['time_factor'],
        row['location_factor']
    ), axis=1
)

print(data)
```
