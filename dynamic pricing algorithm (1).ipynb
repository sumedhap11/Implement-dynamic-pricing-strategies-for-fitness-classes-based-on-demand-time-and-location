```python
import pandas as pd
import numpy as np


df = pd.read_csv("Classes April-May 2018.csv")

df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

# Update column names if needed
class_col = "class_name"
time_col = "class_time"
date_col = "date"
demand_col = "number_booked"
price_col = "price"

# Convert date
df[date_col] = pd.to_datetime(df[date_col])
df["day_of_week"] = df[date_col].dt.day_name()


max_demand = df[demand_col].max()
df["demand_score"] = df[demand_col] / max_demand


# Step 2: Class Popularity Score
# ---------------------------------

class_avg = df.groupby(class_col)[demand_col].transform("mean")
df["popularity_score"] = class_avg / class_avg.max()

# ---------------------------------
# Step 3: Elasticity (Assume value)
# Replace with your model output
# ---------------------------------

elasticity = -0.6   # example from regression

# ---------------------------------
# Step 4: Define Pricing Rules
# ---------------------------------

def dynamic_pricing(row):
    
    base_price = row[price_col]
    demand_factor = 0.20 * row["demand_score"]
    popularity_factor = 0.10 * row["popularity_score"]
    
    # Peak Time Rule
    if row[time_col] in ["6 PM", "7 PM", "8 PM"]:
        time_factor = 0.10
    else:
        time_factor = -0.05
    
    # Weekend Rule
    if row["day_of_week"] in ["Saturday", "Sunday"]:
        weekend_factor = 0.10
    else:
        weekend_factor = 0
    
    # Elasticity adjustment
    elasticity_factor = abs(elasticity) * 0.05
    
    new_price = base_price * (
        1 + demand_factor + popularity_factor + 
        time_factor + weekend_factor - elasticity_factor
    )
    
    return round(new_price, 2)

df["dynamic_price"] = df.apply(dynamic_pricing, axis=1)

print(df[[class_col, demand_col, price_col, "dynamic_price"]].head())
# ---------------------------------
# Revenue Before
# ---------------------------------
df["revenue_before"] = df[price_col] * df[demand_col]

# ---------------------------------
# Revenue After
# ---------------------------------
df["revenue_after"] = df["dynamic_price"] * df[demand_col]

total_before = df["revenue_before"].sum()
total_after = df["revenue_after"].sum()

improvement = ((total_after - total_before) / total_before) * 100

print("Revenue Before:", total_before)
print("Revenue After:", total_after)
print("Revenue Improvement %:", round(improvement, 2))

```
