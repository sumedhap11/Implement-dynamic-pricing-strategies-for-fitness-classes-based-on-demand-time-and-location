```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#load Dataset
df=pd.read_csv("Classes April-May 2018.csv")
print("Dataset Shape:", df.shape)
print(df.head())
print("\nDataset Info:\n", df.info())

#Data Cleaning Checks
print("\nMissing Values :\n",df.isnull().sum())


#Remove duplicate 
df=df.drop_duplicates()

if 'date' in df.columns:
    df['date']=pd.to_datetime(df['date'])
                              
#Descriptive Statistics
print("\nStatistical Summary:\n",df.describe())

if 'price' in df.columns:
    print("Average Price:",df['price'].mean())
if 'number_Booked' in df.columns:
    print("Average Booking:",df['number_Booked'].mean())
if 'revenue' in df.columns:
    print("Revenue Summary:\n",df['Revenue'].describe())                          
                                                        
#Booking Trends Over Time

if 'date' in df.columns and 'number_Booked' in df.columns:
    plt.figure(figsize=(8,5))
    sns.lineplot(x='date',y='number_Booked', data=df)
    plt.title("Booking Trends over Time")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

#price vs Bookings (demand Analysis)
if'price' in df.columns and 'number_Booked' in df.columns:
    plt.figure(figsize=(6,4))
    sns.scatterplot(x='price',y='number_Booked',data=df)
    plt.title("Price vs Bookings")
    plt.tight_layout()
    plt.show()

if 'class_time' in df.columns and 'nimber_Booked' in df.columns:
    plt.figure(figsize=(6,4))
    sns.barplot(x='class_time',y='number_Booked',data=df)
    plt.title("Booking by Time Slot")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

#Location Based Demand
if'location' in df.columns and 'number_Booked' in df.columns:
    plt.figure(figsize=(6,4))
    sns.barplot(x='location', y='number_of_bookings',data=df)
    plt.title("Bookings by Location")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

#Outlier Detection
if 'price' in df.columns:
    plt.figure(figsize=(6,4))
    sns.boxplot(x=df['price'])
    plt.title("Price Distribution ")
    plt.tight_layout()
    plt.show()

if 'number_of_bookings' in df.columns:
    plt.figure(figsize=(6,4))
    sns.boxplot(x=df['number_Booked'])
    plt.title("Booking Distribution")
    plt.tight_layout()
    plt.show()

#Correlation Matrix
numeric_df = df.select_dtypes(include=np.number)

if not numeric_df.empty:
    plt.figure(figsize=(8,6))
    sns.heatmap(numeric_df.corr(),annot=True,cmap='coolwarm')
    plt.title("Feature correlation matrix")
    plt.tight_layout()
    plt.show()

print("\nEDA Successfully!")
                              
```
