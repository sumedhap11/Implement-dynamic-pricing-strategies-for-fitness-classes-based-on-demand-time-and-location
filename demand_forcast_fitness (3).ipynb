```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error
from statsmodels.tsa.arima.model import ARIMA

df = pd.read_csv("Classes April-May 2018.csv")

# Clean column names
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

print("Available Columns:")
print(df.columns)

# Replace these if needed
date_col = "date"
demand_col = "number_booked"

# Convert date column to datetime
df[date_col] = pd.to_datetime(df[date_col])

# Sort by date
df = df.sort_values(by=date_col)

# Set date as index
df.set_index(date_col, inplace=True)

# Keep only demand column
ts = df[demand_col]

# Remove missing values
ts = ts.dropna()

print("\nTime Series Data Shape:", ts.shape)


# 2. TRAIN-TEST SPLIT (Time Based)


train_size = int(len(ts) * 0.8)
train = ts[:train_size]
test = ts[train_size:]


# 3. BUILD ARIMA MODEL
# (p,d,q) can be tuned — using basic (1,1,1)
model = ARIMA(train, order=(1,1,1))
model_fit = model.fit()

# Forecast
forecast = model_fit.forecast(steps=len(test))

# 4. VALIDATE MODEL
mae = mean_absolute_error(test, forecast)
mse = mean_squared_error(test, forecast)

print("\nForecast Model Performance:")
print("Mean Absolute Error (MAE):", round(mae, 2))
print("Mean Squared Error (MSE):", round(mse, 2))


# 5. PLOT FORECAST

plt.figure(figsize=(10,5))
plt.plot(train.index, train, label="Train Data")
plt.plot(test.index, test, label="Actual Demand")
plt.plot(test.index, forecast, label="Forecasted Demand")
plt.title("Demand Forecast using ARIMA")
plt.xlabel("Date")
plt.ylabel("Number of Bookings")
plt.legend()
plt.tight_layout()
plt.show()

print("\nDemand Forecasting Completed Successfully!")
```
